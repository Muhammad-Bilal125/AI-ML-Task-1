```python
# Task 1: Exploring and Visualizing a Simple Dataset

# Import required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Iris dataset using seaborn
iris = sns.load_dataset("iris")

# -------------------------------
# 1. Inspect the Dataset
# -------------------------------

# Print dataset shape
print("Dataset Shape:")
print(iris.shape)

# Print column names
print("\nColumn Names:")
print(iris.columns)

# Print first few rows
print("\nFirst Five Rows:")
print(iris.head())

# Dataset information
print("\nDataset Info:")
print(iris.info())

# Summary statistics
print("\nSummary Statistics:")
print(iris.describe())

# -------------------------------
# 2. Scatter Plot
# -------------------------------

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=iris,
    x="sepal_length",
    y="sepal_width",
    hue="species"
)

plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()

# Another scatter plot for petal features
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Scatter Plot of Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()

# -------------------------------
# 3. Histograms
# -------------------------------

iris.hist(figsize=(10, 8), bins=20)
plt.suptitle("Histograms of Iris Dataset Features")
plt.show()

# Histograms using seaborn
features = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

for feature in features:
    plt.figure(figsize=(7, 5))
    sns.histplot(data=iris, x=feature, kde=True)
    plt.title(f"Distribution of {feature}")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.show()

# -------------------------------
# 4. Box Plots
# -------------------------------

plt.figure(figsize=(10, 6))
sns.boxplot(data=iris)
plt.title("Box Plot of Iris Dataset Features")
plt.show()

# Box plots by species
for feature in features:
    plt.figure(figsize=(8, 5))
    sns.boxplot(data=iris, x="species", y=feature)
    plt.title(f"Box Plot of {feature} by Species")
    plt.xlabel("Species")
    plt.ylabel(feature)
    plt.show()
```
